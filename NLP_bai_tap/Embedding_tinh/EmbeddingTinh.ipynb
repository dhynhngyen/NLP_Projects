```python
# Bộ từ vựng dùng chung cho cả hai phương pháp
vocab = ["tôi", "thích", "học", "AI", "NLP"]

# 1. ONE-HOT ENCODING
def one_hot(word, vocab):
    vector = []

    for v in vocab:
        if v == word:
            vector.append(1)
        else:
            vector.append(0)

    return vector


print("===== ONE-HOT ENCODING =====")

for word in vocab:
    print(word, "->", one_hot(word, vocab))


# 2. BAG OF WORDS (BoW)

def bag_of_words(sentence, vocab):
    # Tách câu thành các từ theo khoảng trắng
    words = sentence.split()

    vector = []

    for v in vocab:
        # Đếm số lần từ v xuất hiện trong câu
        count = words.count(v)
        vector.append(count)

    return vector


sentences = [
    "tôi thích học AI",
    "tôi thích học NLP",
    "tôi thích AI AI"
]

print("\n===== BAG OF WORDS =====")

for sentence in sentences:
    print(sentence, "->", bag_of_words(sentence, vocab))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Tập hợp các văn bản
documents = [
    "tôi thích học AI",
    "tôi thích học NLP",
    "tôi thích AI AI"
]

# Khởi tạo bộ chuyển đổi văn bản sang TF-IDF
vectorizer = TfidfVectorizer()

# Học bộ từ vựng, tính IDF và chuyển đổi văn bản
tfidf_matrix = vectorizer.fit_transform(documents)

# Lấy danh sách từ và thứ tự cột trong ma trận
feature_names = vectorizer.get_feature_names_out()

print("Bộ từ vựng:")
print(feature_names)

print("\nMa trận TF-IDF:")
print(tfidf_matrix.toarray().round(3))

print("\nTrọng số IDF của từng từ:")
for word, idf in zip(feature_names, vectorizer.idf_):
    print(f"{word}: {idf:.3f}")
```

```python
from gensim.models import Word2Vec

# 1. Tập câu đã được tách thành từ
sentences = [
    ["cat", "is", "a", "pet"],
    ["dog", "is", "a", "pet"],
    ["cat", "likes", "fish"],
    ["dog", "likes", "meat"],
    ["kitten", "is", "a", "cat"],
    ["puppy", "is", "a", "dog"],
    ["cat", "and", "dog", "are", "pets"],
    ["fish", "is", "food"],
    ["dog", "eats", "meat"],
    ["cat", "eats", "fish"]
]

# 2. Khởi tạo và huấn luyện mô hình
w2v_model = Word2Vec(
    sentences=sentences,
    vector_size=20,
    window=2,
    min_count=1,
    sg=1,
    workers=1,
    epochs=100,
    seed=42
)

# 3. Lấy vector của một từ đã học
word = "cat"
vector = w2v_model.wv[word]

print("Vector của cat:")
print(vector)

print("Kích thước vector:", vector.shape)

# 4. Tìm các từ gần cat nhất
print("\nCác từ gần cat:")
print(w2v_model.wv.most_similar("cat", topn=5))

# 5. Kiểm tra một từ chưa có trong bộ từ vựng
new_word = "cats"

print("\nWord2Vec có từ cats không?")
print(new_word in w2v_model.wv.key_to_index)
print(w2v_model.wv.key_to_index)

# 6. Thử truy xuất vector của từ chưa có
if new_word in w2v_model.wv.key_to_index:
    print(w2v_model.wv[new_word])
else:
    print("Word2Vec thông thường không có vector cho từ này.")
```

```python
import gensim.downloader as api

# 1. Tải và nạp bộ vector GloVe đã huấn luyện sẵn
glove_model = api.load("glove-wiki-gigaword-50")

# 2. Lấy vector của một từ đã có trong bộ từ vựng
word = "computer"
vector = glove_model[word]

print("10 chiều đầu của vector computer:")
print(vector[:10])

print("Kích thước vector:", vector.shape)

# 3. Tìm những từ gần computer
print("\nCác từ gần computer:")
print(glove_model.most_similar("computer", topn=5))

# 4. Kiểm tra từ mới có trong bộ từ vựng không
new_word = "quorblif"

print("\nGloVe có từ quorblif không?")
print(new_word in glove_model.key_to_index)

# 5. Chỉ lấy vector nếu từ đã có trong bộ từ vựng
if new_word in glove_model.key_to_index:
    print(glove_model[new_word])
else:
    print("GloVe không có vector đã học cho từ này.")
```

```python
from gensim.models import FastText

# 1. Tập câu đã được tách thành từ
sentences = [
    ["cat", "is", "a", "pet"],
    ["dog", "is", "a", "pet"],
    ["cat", "likes", "fish"],
    ["dog", "likes", "meat"],
    ["kitten", "is", "a", "cat"],
    ["puppy", "is", "a", "dog"],
    ["cat", "and", "dog", "are", "pets"],
    ["fish", "is", "food"],
    ["dog", "eats", "meat"],
    ["cat", "eats", "fish"]
]

# 2. Khởi tạo và huấn luyện FastText
fasttext_model = FastText(
    sentences=sentences,
    vector_size=20,
    window=2,
    min_count=1,
    sg=1,
    min_n=3,
    max_n=6,
    workers=1,
    epochs=100,
    seed=42
)

# 3. Lấy vector của từ đã xuất hiện
print("Vector của cat:")
print(fasttext_model.wv["cat"])

# 4. Kiểm tra từ cats có nằm trong bộ từ vựng không
new_word = "cats"

print("\nFastText có cats trong bộ từ vựng không?")
print(new_word in fasttext_model.wv.key_to_index)

# 5. Thử tạo vector cho cats dù từ không nằm trong bộ từ vựng
new_vector = fasttext_model.wv[new_word]

print("\nVector của cats:")
print(new_vector)

print("Kích thước vector:", new_vector.shape)

# 6. So sánh vector cats với cat
print("\nĐộ tương đồng giữa cats và cat:")
print(fasttext_model.wv.similarity("cats", "cat"))
```
