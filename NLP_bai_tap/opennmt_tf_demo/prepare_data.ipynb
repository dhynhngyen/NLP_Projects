```python
from pathlib import Path
from datasets import load_dataset

DATA_DIR = Path("data")
DATA_DIR.mkdir(parents=True,exist_ok=True)

dataset = load_dataset("ncduy/mt-en-vi")

print(dataset)
```

```python
def clean_text(text):
    if not isinstance(text,str):
        return ""

    return " ".join(text.split())

def save_parallel_data(split_name, source_file, target_file, limit=None,shuffle=False):
    split = dataset[split_name]
    if shuffle:
        split = split.shuffle(seed=42)
    if limit is not None:
        split = split.select(range(min(limit, len(split))))

    count = 0
    source_path = DATA_DIR/ source_file
    target_path = DATA_DIR/ target_file

    with open(source_path, "w", encoding='utf-8') as source_writer, \
    open(target_path,"w",encoding="utf-8") as target_writer:
        for example in split:
            en = clean_text(example.get("en"))
            vi = clean_text(example.get("vi"))

            if not en or not vi:
                continue

            source_writer.write(en + "\n")
            target_writer.write(vi + "\n")

            count += 1

        print(f"{split_name}: đã ghi {count} cặp câu. ")
        
```

```python
save_parallel_data(
    split_name='train',
    source_file='train.en',
    target_file="train.vi",
    limit=50000,
    shuffle=True    
)
```

```python
save_parallel_data(split_name='validation',
                   source_file='valid.en',
                   target_file='valid.vi')

save_parallel_data(
    split_name='test',
    source_file='test.en',
    target_file='test.vi'   
)
```
